# Project 1. AI Text Summarizer

````python
from google.colab import userdata
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel, RunnablePassthrough

from google.colab import userdata
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="nvidia/nemotron-3.5-lightning-30b-a3b",
    base_url="https://integrate.api.nvidia.com/v1",
    api_key=userdata.get("NVIDIA_API_KEY"),
    temperature=0.2,
    max_tokens=2048
)
````

````python
prompt = ChatPromptTemplate.from_template("""
You are an expert text summarizer.

Summarize the following article in simple language.

Requirements:
- Identify the main idea.
- Include important facts.
- Remove unnecessary details.
- Use clear bullet points.
- End with a one-sentence conclusion.

ARTICLE:
{article}
""")
````

````python
summary_chain = prompt | llm | StrOutputParser()

chain = RunnableParallel({
    "original_article": RunnablePassthrough(),
    "summary": summary_chain
})
````

````python
article = input("Enter the article to summarize: ")

result = chain.invoke({
    "article": article
})

print("\n" + "=" * 60)
print("AI TEXT SUMMARIZER")
print("=" * 60)

print("\nORIGINAL ARTICLE:")
print(result["original_article"]["article"])

print("\nGENERATED SUMMARY:")
print(result["summary"])
````

# Project 2.Resume Analyzer









Objective: Analyze a resume and identify skills, experience, education, strengths, and improvement areas.
Pipeline: Resume → Skills + Experience + Education → Suggestions

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableLambda

prompt = ChatPromptTemplate.from_template("""
You are an experienced technical resume reviewer.

Analyze the resume below and provide a detailed report.

1. SKILLS
   - Technical skills
   - Programming languages
   - Frameworks and tools

2. EXPERIENCE
   - Work experience and internships
   - Projects
   - Responsibilities and achievements

3. EDUCATION
   - Degrees and institutions
   - Academic details

4. STRENGTHS
   - Strong points in the resume

5. MISSING INFORMATION
   - Important details that are unclear or absent

6. IMPROVEMENT SUGGESTIONS
   - Specific improvements
   - ATS-friendly formatting
   - Relevant project and achievement improvements

Do not invent qualifications, experience, or achievements.
Clearly identify missing information.

RESUME:
{resume}
""")

# Extract the resume string, run the model, and parse its output
analysis_chain = (
    RunnableLambda(lambda x: {"resume": x["resume"]})
    | prompt
    | llm
    | StrOutputParser()
)

# Preserve the input and add the analysis
chain = RunnablePassthrough.assign(
    analysis=analysis_chain
)

resume = input("Paste your resume:\n")

result = chain.invoke({"resume": resume})

print("\nRESUME ANALYSIS")
print("=" * 50)
print("Resume characters:", len(result["resume"]))
print("\nDETAILED REPORT:\n")

if result["analysis"] and result["analysis"].strip():
    print(result["analysis"])
else:
    print("The model returned an empty response. Check the model/API output.")
````

# Project 3.Interview Question Generator
Objective: Generate beginner, intermediate, and advanced interview questions based on the skills provided by the user.

````python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

prompt = ChatPromptTemplate.from_template("""
You are an experienced technical interviewer.

Generate interview questions for these skills:
{skills}

Create three sections:

BEGINNER
- Generate 5 foundational questions.

INTERMEDIATE
- Generate 5 practical questions.
- Include coding or application-based questions where appropriate.

ADVANCED
- Generate 5 challenging questions.
- Include system design, optimization, or deeper concepts where relevant.

For every question:
- Give a clear question number.
- Include a short hint.
- Avoid repeating the same question.
- Match the difficulty level to the section.

Do not provide full answers unless requested.
""")

chain = RunnablePassthrough.assign(
    questions=prompt | llm | parser
)

skills = input(
    "Enter skills separated by commas "
    "(e.g. Python, SQL, Machine Learning): "
)

data = {"skills": skills}

result = chain.invoke(data)

print("\nINTERVIEW QUESTION GENERATOR")
print("=" * 50)
print("Skills:", result["skills"])
print("\n" + result["questions"])
````

# Project 4.Professional Email Generator

````python

prompt = ChatPromptTemplate.from_template("""
You are a professional email-writing assistant.

Write an email using these details:

Purpose: {purpose}
Tone: {tone}
Recipient type: {recipient_type}

Instructions:
- Generate a relevant subject line.
- Use an appropriate greeting.
- Explain the purpose clearly.
- Include a polite and professional closing.
- Keep the email concise but complete.
- Do not invent names, dates, or commitments.
- Return only the subject and email draft.

""")
chain=RunnablePassthrough.assign(
    email=prompt | llm | parser
)

purpose = input("Enter the purpose of the email: ")
tone = input("Enter the tone of the email: ")
recipient_type = input("Enter the recipient type: ")

data = {
    "purpose": purpose,
    "tone": tone,
    "recipient_type": recipient_type
}

result = chain.invoke(data)

print("\nGENERATED EMAIL")
print("=" * 50)
print("Purpose:", result["purpose"])
print("Tone:", result["tone"])
print("Recipient:", result["recipient_type"])
print("\n" + result["email"])
````

````python

prompt = ChatPromptTemplate.from_template("""
You are an expert SQL developer.

DATABASE TYPE: {database_type}
TABLE SCHEMA:
{schema}

USER REQUEST:
{request}

Instructions:
1. Generate a SQL query for the given request.
2. Use only the tables and columns supplied in the schema.
3. Use the specified database dialect.
4. Explain important assumptions.
5. If the request cannot be answered using the schema,
   explain which information is missing.
6. Prefer a read-only SELECT query.
7. Never generate destructive statements such as
   DROP, DELETE, UPDATE, or TRUNCATE.
8. Return the SQL query in a SQL code block followed
   by a short explanation.

Do not claim that the query was executed.
""")

chain = RunnablePassthrough.assign(
    generated_sql=prompt | llm | parser
)

database_type = input(
    "Database (MySQL, PostgreSQL, SQLite): "
)

schema = input(
    "Enter table schema, e.g. "
    "employees(id, name, salary, department): "
)

request = input("Describe what you want in plain English: ")

data = {
    "database_type": database_type,
    "schema": schema,
    "request": request
}

result = chain.invoke(data)

print("\nSQL QUERY GENERATOR")
print("=" * 50)
print("Request:", result["request"])
print("Database:", result["database_type"])
print("\nGenerated SQL:\n")
print(result["generated_sql"])
````
