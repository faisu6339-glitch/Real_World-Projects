Complete Code (1 Image per Class + All Preprocessing)

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

base_path = "/content/drive/MyDrive/main_dataset/dataset_sir/main_dataset/sorted_dataset"
classes = ["CBB", "CBSD", "CGM", "CMD", "Healthy"]

for cls in classes:
    class_path = os.path.join(base_path, cls)

    # pick first image
    img_name = os.listdir(class_path)[0]
    img_path = os.path.join(class_path, img_name)

    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # --- PREPROCESSING ---

    # Resize
    resized = cv2.resize(img, (224, 224))

    # Normalization
    normalized = resized / 255.0

    # Grayscale
    gray = cv2.cvtColor(resized, cv2.COLOR_RGB2GRAY)

    # Enhancement (brightness/contrast)
    enhanced = cv2.convertScaleAbs(resized, alpha=1.5, beta=30)

    # Blur
    blur = cv2.GaussianBlur(resized, (5,5), 0)

    # Edge Detection
    edges = cv2.Canny(gray, 100, 200)

    # Sharpen
    kernel = np.array([[0,-1,0],
                       [-1,5,-1],
                       [0,-1,0]])
    sharpen = cv2.filter2D(resized, -1, kernel)

    # Histogram Equalization
    hist_eq = cv2.equalizeHist(gray)

    # --- AUGMENTATION ---

    # Flip
    flip = cv2.flip(resized, 1)

    # Rotation
    (h, w) = resized.shape[:2]
    center = (w // 2, h // 2)
    matrix = cv2.getRotationMatrix2D(center, 25, 1.0)
    rotated = cv2.warpAffine(resized, matrix, (w, h))

    # Zoom (crop + resize)
    zoom = resized[50:200, 50:200]
    zoom = cv2.resize(zoom, (224,224))

    # --- Plot ---

    images = [
        resized, normalized, gray, enhanced,
        blur, edges, sharpen, hist_eq,
        flip, rotated, zoom
    ]

    titles = [
        "Resized", "Normalized", "Gray", "Enhanced",
        "Blur", "Edges", "Sharpen", "Hist Eq",
        "Flipped", "Rotated", "Zoomed"
    ]

    plt.figure(figsize=(18,10))

    for i in range(len(images)):
        plt.subplot(3,4,i+1)

        if len(images[i].shape) == 2:
            plt.imshow(images[i], cmap='gray')
        else:
            plt.imshow(images[i])

        plt.title(titles[i])
        plt.axis("off")

    plt.suptitle(f"Class: {cls}", fontsize=16)
    plt.tight_layout()
    plt.show()
```

Train

```python
import tensorflow as tf

train_data = tf.keras.utils.image_dataset_from_directory(
    data_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(224,224),
    batch_size=32
)
```

Validation

```python
val_data = tf.keras.utils.image_dataset_from_directory(
    data_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(224,224),
    batch_size=32
)
```

preprocessing

```python
from tensorflow.keras.applications.mobilenet_v3 import preprocess_input

train_data = train_data.map(lambda x, y: (preprocess_input(x), y))
val_data   = val_data.map(lambda x, y: (preprocess_input(x), y))
```

Data Augmentation

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.2),
    tf.keras.layers.RandomZoom(0.2),
])

train_data = train_data.map(lambda x, y: (data_augmentation(x), y))
```

Optimize Dataset

```python
AUTOTUNE = tf.data.AUTOTUNE

train_data = train_data.cache().shuffle(1000).prefetch(AUTOTUNE)
val_data   = val_data.cache().prefetch(AUTOTUNE)
```

Build Mobilenetv3 small Model (TensorFlow / Keras)

```python
base_model = tf.keras.applications.MobileNetV3Small(
    input_shape=(224,224,3),
    include_top=False,
    weights='imagenet'
)

base_model.trainable = False
```

```python
x = base_model.output
x = tf.keras.layers.GlobalAveragePooling2D()(x)
x = tf.keras.layers.BatchNormalization()(x)
x = tf.keras.layers.Dense(256, activation='relu')(x)
x = tf.keras.layers.Dropout(0.5)(x)

output = tf.keras.layers.Dense(5, activation='softmax')(x)

model = tf.keras.Model(inputs=base_model.input, outputs=output)
```

```python
model.summary()
```

compile

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
callbacks = [
    tf.keras.callbacks.EarlyStopping(patience=5, restore_best_weights=True),
    tf.keras.callbacks.ReduceLROnPlateau(patience=3)
]

history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=20,
    callbacks=callbacks
)
```

```python
print("Final Training Accuracy:", history.history['accuracy'][-1])
print("Final Validation Accuracy:", history.history['val_accuracy'][-1])
```

```python
import matplotlib.pyplot as plt

# Accuracy graph
plt.plot(history.history['accuracy'], label='train_acc')
plt.plot(history.history['val_accuracy'], label='val_acc')
plt.legend()
plt.title("Accuracy")
plt.show()

# Loss graph
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.legend()
plt.title("Loss")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
y_true = []
y_pred = []

for images, labels in val_data:
    preds = model.predict(images)

    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(preds, axis=1))
```

Single Image Prediction

```python
import numpy as np
from tensorflow.keras.preprocessing import image
import matplotlib.pyplot as plt
import os

# Image path
# Find an existing image path. For example, taking the first image from the 'CBB' class.
# Ensure 'data_path' and 'classes' variables are available from previous cells.

if 'data_path' in locals() and 'classes' in locals() and len(classes) > 0:
    # Assuming 'CBB' is one of the classes, and data_path is set correctly
    class_to_use = 'CBB'
    class_path = os.path.join(data_path, class_to_use)
    if os.path.exists(class_path):
        image_files = os.listdir(class_path)
        if image_files:
            img_path = os.path.join(class_path, image_files[0])
        else:
            print(f"No images found in {class_path}. Please check the directory.")
            img_path = ""
    else:
        print(f"Class path {class_path} does not exist. Please check data_path and class_name.")
        img_path = ""
else:
    print("Variables data_path or classes not defined. Please ensure previous cells are run.")
    img_path = ""

# Only proceed if a valid img_path was found
if img_path:
    # Load image
    img = image.load_img(img_path, target_size=(224,224))

    # Show image
    plt.imshow(img)
    plt.axis('off')
    plt.show()

    # Convert to array
    img_array = image.img_to_array(img)

    # Batch dimension add
    img_array = np.expand_dims(img_array, axis=0)

    # (Optional but recommended) Normalize
    img_array = img_array / 255.0

    # Prediction
    pred = model.predict(img_array)

    # Class names
    # Use the 'classes' variable which holds the list of class names from the dataset
    class_names = classes

    # Result
    pred_index = np.argmax(pred)
    confidence = np.max(pred)

    print("Prediction:", class_names[pred_index])
    print("Confidence:", confidence)
else:
    print("Image path is not valid. Cannot proceed with prediction.")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image

folder_path = "test_images"

for img_name in os.listdir(folder_path):
    img_path = os.path.join(folder_path, img_name)

    # Load image
    img = image.load_img(img_path, target_size=(224,224))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) / 255.0

    # Prediction
    pred = model.predict(img_array)[0]

    pred_class = class_names[np.argmax(pred)]
    confidence = np.max(pred)

    # Show image with prediction
    plt.imshow(img)
    plt.title(f"{pred_class} ({confidence:.2f})")
    plt.axis('off')
    plt.show()

    # Optional: print all class probabilities
    print(f"\nImage: {img_name}")
    for i, class_name in enumerate(class_names):
        print(f"{class_name}: {pred[i]:.4f}")
```
